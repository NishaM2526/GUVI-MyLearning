## Write a function that calculate factorial of a number. Number is provided as argument

```python
def factorial(num):
    index=1
    fact=1
    while index<=num:
        fact=fact*index
        index=index+1
    return fact

factorial(5)
```

## To write a function that will count number of digits. Eg:9432 (4 digits)

```python
def digit_length(num):
    count=1
    index=num

    while index!=0:
        index=index//10 
        if index!=0 :
            count=count+1
        print(index)

    print(f"Total number of digits in {num} is {count}")

digit_length(222)
```

# Exercise 3 :
## Find all the perfect square in range 10 to 40

```python
import math

def perfect_square(num1,num2):
    squares=[]  
    index=num1 
    while index <= num2:
        sqr_root=math.sqrt(index)
        value=math.trunc(sqr_root)
        if value**2 == index:
            print("Its a square")
            squares.append(index)
        index=index+1
    print(squares)

perfect_square(10,40)
```

```python
import math

def perfect_square(num1,num2):
    squares=[]  
    for i in range(num1,num2) :
        sqr_root=math.sqrt(i)
        value=math.trunc(sqr_root)
        if value**2 == i:
            print("Its a square")
            squares.append(i)
    print(squares)

perfect_square(10,20)
```
