# Exercise 1 :
## Write a function that calculate factorial of a number. Number is provided as argument

```python
def factorial(num):
    index=1
    fact=1
    while index<=num:
        fact=fact*index
        index=index+1
    return fact

factorial(5)
```

# Exercise 3 :
## Find all the perfect square in range 10 to 40

```python
import math

def perfect_square(num1,num2):
    squares=[]  
    index=num1 
    while index <= num2:
        sqr_root=math.sqrt(index)
        value=math.trunc(sqr_root)
        if value**2 == index:
            print("Its a square")
            squares.append(index)
        index=index+1
    print(squares)

perfect_square(10,40)
```

```python
import math

def perfect_square(num1,num2):
    squares=[]  
    for i in range(num1,num2) :
        sqr_root=math.sqrt(i)
        value=math.trunc(sqr_root)
        if value**2 == i:
            print("Its a square")
            squares.append(i)
    print(squares)

perfect_square(10,100)
```
