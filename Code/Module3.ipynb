```python
# Binary search

def create_node(value):
    return{
        'key':value,
        'left':None,
        'right':None
    }

def bst_insert(root,key):
    if root is None:
        return create_node(key)
    
    if key < root['key']:
        root['left'] = bst_insert(root['left'],key)
    elif key > root['key']:
        root['right'] = bst_insert(root['right'],key)

    return root

values=[50,30,60,70,40,80]

root=None
for v in values:
    root=bst_insert(root,v)
print(root)
```
