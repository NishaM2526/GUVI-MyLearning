## Consider a list having scores of 10 learners. Create Variable V percentage to calculate percentage of individual learners. Max marks = 130

```python
learners_scores = [70,35,120,65,82,40,130,98,55,110]
learners_perc = []
max_mark = 130

for var in learners_scores:
    v = var/max_mark 
    perc = round(v * 100,2)
    learners_perc.append(perc)

print(f"Learners Score : {learners_scores}")
print(f"Learners Percentange : {learners_perc}")
```
